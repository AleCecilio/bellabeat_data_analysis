## 1. Importação das Bibliotecas

```python
# Configuração do Jupyter (Autoreload)
%load_ext autoreload
%autoreload 2

# Imports 
import os
import sys
from pathlib import Path
import pandas as pd

# Adiciona a raiz do projeto ao sys.path para conseguir importar a pasta src
sys.path.append(os.path.abspath('..'))

from src.data_loader import inspect_samples, load_and_concatenate_csvs

RAW_DATA_PATH = "../data/raw"
```

---

## 2. Inspeção Inicial

```python
# Carrega amostras de 3 linhas de todas as tabelas
samples = inspect_samples(RAW_DATA_PATH, n_rows=3)

print(f"Total de tabelas encontradas no dataset: {len(samples)}\n")

# Exibe as colunas e as primeiras linhas de cada tabela para tomada de decisão
for table_name, df_sample in samples.items():
    print(f"=== TABELA: {table_name} ===")
    print(f"Colunas ({len(df_sample.columns)}): {list(df_sample.columns)}")
    display(df_sample)
    print("\n" + "-" * 60 + "\n")
```

---

## 3. Seleção de Tabelas

```python
# Selecao baseada na exploracao
tabelas_selecionadas = [
    "dailyActivity_merged",
    "sleepDay_merged",
    "weightLogInfo_merged",
    "hourlySteps_merged"
]
```

---

## 4. Concatenação dos Datasets Selecionados

```python
# Carregamento e Concatenacao (usando o loop da funcao do src
dfs = {} # Dicionario para armazenar nossos DataFrames

for nome_tabela in tabelas_selecionadas:
    print(f"Carregando e concatenando: {nome_tabela}...")
    dfs[nome_tabela] = load_and_concatenate_csvs(RAW_DATA_PATH, nome_tabela)

print("\nCarga finalizada!")
```

--- 

## 5. Sanity Check Pós-concatenação

```python
# Relatório pós-concatenação
for table_name, df in dfs.items():
    print(f"=== {table_name.upper()} ===")
    print(f"Volume: {df.shape[0]} linhas e {df.shape[1]} colunas")
    
    # Exibe os tipos de dados que o Pandas identificou automaticamente
    print("Tipos de Dados:")
    display(df.dtypes.to_frame(name="Tipo").T)
    
    # Exibe as 3 primeiras linhas para checar se a estrutura faz sentido
    display(df.head(3))
    print("\n" + "="*80 + "\n")
```

---

## 6. Salvamento dos Dados Consolidados

```python
# Garante que a pasta de destino exista
os.makedirs("../data/processed/bronze", exist_ok=True)

# Exporta cada dataframe concatenado
for table_name, df in dfs.items():
    # O arquivo será salvo na camada intermediária
    save_path = f"../data/processed/bronze/{table_name}_concat.csv"
    
    # index=False evita criar uma coluna inútil com o número da linha
    df.to_csv(save_path, index=False)
    print(f"Ingestão finalizada: {save_path} ({df.shape[0]} linhas geradas)")

print("\nIngestão completa! Os dados estão prontos para Profiling e Limpeza no Notebook 02.")
```
